```python
import pandas as pd

orders = pd.read_csv("Day9_Orders.csv")
customers = pd.read_csv("Day9_Customers.csv")
products = pd.read_csv("Day9_Products.csv")

print("Orders:")
display(orders.head())

print("\nCustomers:")
display(customers.head())

print("\nProducts:")
display(products.head())
```

```python
print("Orders columns:")
print(orders.columns.tolist())

print("\nCustomers columns:")
print(customers.columns.tolist())

print("\nProducts columns:")
print(products.columns.tolist())

print("\nDataset shapes:")
print("Orders:", orders.shape)
print("Customers:", customers.shape)
print("Products:", products.shape)
```

## Merging Orders and Customers

```python
orders_customers = pd.merge(
    orders,
    customers,
    on="Customer_ID",
    how="left"
)

print("Merged Orders + Customers:")
display(orders_customers.head())
```

```python
print("Merged dataset shape:", orders_customers.shape)
print("Merged dataset columns:")
print(orders_customers.columns.tolist())
```

## Merging Products

```python
processed_df = pd.merge(
    orders_customers,
    products,
    on="Product_ID",
    how="left"
)

print("Orders + Customers + Products:")
display(processed_df.head())
```

```python
print("Processed dataset shape:", processed_df.shape)

print("\nProcessed dataset columns:")
print(processed_df.columns.tolist())
```

## Concatenating DataFrames

```python
df_part1 = orders.head(5)
df_part2 = orders.tail(5)

print("First part:")
display(df_part1)

print("Second part:")
display(df_part2)
```

```python
orders_combined = pd.concat(
    [df_part1, df_part2],
    ignore_index=True
)

print("Concatenated DataFrame:")
display(orders_combined)
```

```python
print("Original Orders rows:", len(orders))
print("Part 1 rows:", len(df_part1))
print("Part 2 rows:", len(df_part2))
print("Concatenated rows:", len(orders_combined))
```

## Creating a Derived Column Using apply()

```python
print(processed_df.columns.tolist())
```

```python
def classify_quantity(quantity):
    if quantity >= 8:
        return "High"
    elif quantity >= 4:
        return "Medium"
    else:
        return "Low"

processed_df["Quantity_Category"] = processed_df["Quantity"].apply(classify_quantity)

display(
    processed_df[
        ["Order_ID", "Product_ID", "Quantity", "Quantity_Category"]
    ].head(10)
)
```

```python
print("Quantity Category Distribution:")
display(processed_df["Quantity_Category"].value_counts())
```

## Date and Time Processing

```python
processed_df["Order_Date"] = pd.to_datetime(processed_df["Order_Date"])

print("Order_Date data type:")
print(processed_df["Order_Date"].dtype)

display(processed_df[["Order_ID", "Order_Date"]].head())
```

```python
processed_df["Order_Date"] = pd.to_datetime(processed_df["Order_Date"])

print("Order_Date data type:")
print(processed_df["Order_Date"].dtype)

display(processed_df[["Order_ID", "Order_Date"]].head())
```

```python
processed_df["Order_Day"] = processed_df["Order_Date"].dt.day

display(
    processed_df[
        ["Order_ID", "Order_Date", "Order_Day"]
    ].head(10)
)
```

```python
processed_df["Order_DayOfWeek"] = processed_df["Order_Date"].dt.day_name()

display(
    processed_df[
        ["Order_ID", "Order_Date", "Order_DayOfWeek"]
    ].head(10)
)
```

```python
processed_df["Order_Month"] = processed_df["Order_Date"].dt.month

display(
    processed_df[
        ["Order_ID", "Order_Date", "Order_Month"]
    ].head(10)
)
```

```python
display(
    processed_df[
        [
            "Order_ID",
            "Order_Date",
            "Order_Month",
            "Order_Day",
            "Order_DayOfWeek"
        ]
    ].head(10)
)
```

## Final Data Cleaning and Inspection

```python
print("Missing values in each column:")
display(processed_df.isnull().sum())
```

```python
print("Rows before removing duplicates:", len(processed_df))

processed_df = processed_df.drop_duplicates()

print("Rows after removing duplicates:", len(processed_df))
```

```python
print("Final dataset shape:", processed_df.shape)

print("\nFinal columns:")
print(processed_df.columns.tolist())

display(processed_df.head(10))
```

```python
processed_df.to_csv(
    "Day9_Processed_Ecommerce.csv",
    index=False
)

print("Processed dataset exported successfully!")
```
