# E-commerce Sales Analysis

This notebook analyzes an e-commerce sales dataset using Python and Pandas.
It explores sales based on categories, cities, products, and payment methods,
along with filtering, sorting, grouping, and aggregation operations.

```python
import pandas as pd
```

```python
df = pd.read_csv("Day8_Ecommerce_Sales_Dataset.csv")

df.head()
```

```python
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)
```

## 2. Basic Sales Statistics

```python
total_sales = df["Total_Sales"].sum()

print("Total Sales:", total_sales)
```

```python
average_sales = df["Total_Sales"].mean()

print("Average Sales per Order:", average_sales)
```

```python
highest_sales = df["Total_Sales"].max()
lowest_sales = df["Total_Sales"].min()

print("Highest Sale:", highest_sales)
print("Lowest Sale:", lowest_sales)
```

```python
total_quantity = df["Quantity"].sum()

print("Total Quantity Sold:", total_quantity)
```

```python
number_of_orders = df["Order_ID"].count()

print("Number of Orders:", number_of_orders)
```

```python
print("Number of Orders:", df["Order_ID"].count())
print("Total Quantity Sold:", df["Quantity"].sum())
print("Total Sales:", df["Total_Sales"].sum())
print("Average Sales per Order:", df["Total_Sales"].mean())
print("Highest Sale:", df["Total_Sales"].max())
print("Lowest Sale:", df["Total_Sales"].min())
```

## 3. Filtering and Sorting

```python
high_value_orders = df[df["Total_Sales"] > 5000]

print("Orders with sales greater than ₹5,000:")
display(high_value_orders)
```

```python
sorted_sales = df.sort_values("Total_Sales", ascending=False)

print("Orders sorted by Total Sales:")
display(sorted_sales[["Order_ID", "Customer", "Product", "Total_Sales"]].head(10))
```

```python
highest_order = df.loc[df["Total_Sales"].idxmax()]

print("Highest-Selling Order:")
display(highest_order)
```

```python
low_value_orders = df[df["Total_Sales"] < 2000]

print("Orders with sales below ₹2,000:")
display(low_value_orders)
```

```python
quantity_sorted = df.sort_values("Quantity", ascending=False)

print("Orders with the highest quantities:")
display(
    quantity_sorted[
        ["Order_ID", "Product", "Quantity", "Total_Sales"]
    ].head(10)
)
```

## 4. Grouping and Aggregation

```python
category_sales = (
    df.groupby("Category")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Total Sales by Category:")
display(category_sales)
```

```python
category_quantity = (
    df.groupby("Category")["Quantity"]
    .sum()
    .sort_values(ascending=False)
)

print("Total Quantity Sold by Category:")
display(category_quantity)
```

```python
city_sales = (
    df.groupby("City")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Total Sales by City:")
display(city_sales)
```

```python
print("Top 10 Cities by Sales:")
display(city_sales.head(10))
```

```python
product_sales = (
    df.groupby("Product")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Total Sales by Product:")
display(product_sales)
```

```python
print("Top 10 Products by Sales:")
display(product_sales.head(10))
```

```python
payment_sales = (
    df.groupby("Payment_Method")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Total Sales by Payment Method:")
display(payment_sales)
```

```python
category_summary = df.groupby("Category").agg(
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Total_Quantity=("Quantity", "sum"),
    Average_Rating=("Rating", "mean")
).sort_values("Total_Sales", ascending=False)

print("Category Summary:")
display(category_summary)
```

## 5. Identifying Top-Performing Categories and Products

```python
top_category = category_sales.idxmax()
top_category_sales = category_sales.max()

print("Top-Performing Category:", top_category)
print("Total Sales:", top_category_sales)
```

```python
lowest_category = category_sales.idxmin()
lowest_category_sales = category_sales.min()

print("Lowest-Performing Category:", lowest_category)
print("Total Sales:", lowest_category_sales)
```

```python
print("Top 5 Products by Total Sales:")
display(product_sales.head(5))
```

```python
print("Lowest 5 Products by Total Sales:")
display(product_sales.tail(5).sort_values())
```

```python
top_city = city_sales.idxmax()
top_city_sales = city_sales.max()

print("Top-Performing City:", top_city)
print("Total Sales:", top_city_sales)
```

```python
top_payment = payment_sales.idxmax()
top_payment_sales = payment_sales.max()

print("Payment Method with Highest Sales:", top_payment)
print("Total Sales:", top_payment_sales)
```

```python
sports_products = df[df["Category"] == "Sports"]

print("Sports Category Orders:")
display(
    sports_products[
        ["Order_ID", "Product", "Quantity", "Total_Sales"]
    ]
)
```

```python
high_rated = df[df["Rating"] >= 4.5]

print("Orders with Rating 4.5 or Higher:")
display(
    high_rated[
        ["Order_ID", "Product", "Category", "Rating", "Total_Sales"]
    ]
)
```

## 6. Key Observations

```python
print(f"1. The top-performing category by total sales is {top_category}.")
print(f"   Total sales: {top_category_sales:.2f}")

print(f"\n2. The lowest-performing category by total sales is {lowest_category}.")
print(f"   Total sales: {lowest_category_sales:.2f}")

print(f"\n3. The top-performing city is {top_city}.")
print(f"   Total sales: {top_city_sales:.2f}")

print(f"\n4. The payment method generating the highest sales is {top_payment}.")
print(f"   Total sales: {top_payment_sales:.2f}")

print("\n5. The top 5 products by total sales are:")
display(product_sales.head(5))
```

## Conclusion

The e-commerce dataset was analyzed using Pandas to understand sales
performance across different categories, cities, products, and payment
methods. The analysis used filtering, sorting, grouping, and aggregation
to identify high-value orders and top-performing areas.

The results provide an overview of sales patterns and can help identify
strong-performing categories, products, cities, and payment methods.
