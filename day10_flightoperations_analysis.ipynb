```python
import pandas as pd

df = pd.read_csv("Day10_Flight_Operations_Dataset.csv")

df.head()
```

```python
print("First 5 rows:")
display(df.head())

print("\nLast 5 rows:")
display(df.tail())
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)
```

```python
df.info()
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
print("Missing Value Percentage:")
print((df.isnull().sum() / len(df) * 100).round(2))
```

```python
print("Statistical Summary:")
df.describe()
```

```python
print("Categorical Column Summary:")
df.describe(include='object')
```

```python
print("Single Column:")
display(df['Flight_Status'])

print("Multiple Columns:")
display(df[['Flight_Status', 'Airline']])

print("First 5 Rows:")
display(df.iloc[0:5])

print("First 5 Rows and First 3 Columns:")
display(df.iloc[0:5, 0:3])

print("First Row:")
display(df.loc[0])
```

```python
print("On Time Flights:")
display(df[df['Flight_Status'] == 'On Time'])

print("Delayed Flights:")
display(df[df['Flight_Status'] == 'Delayed'])

print("Cancelled Flights:")
display(df[df['Flight_Status'] == 'Cancelled'])

print("Flight Status Counts:")
display(df['Flight_Status'].value_counts())
```

```python
print("Sorted by Airline:")
display(df.sort_values(by='Airline'))

print("Sorted by Airline (Descending):")
display(df.sort_values(by='Airline', ascending=False))

print("Sorted by Flight Status:")
display(df.sort_values(by='Flight_Status'))

print("Sorted by Airline and Flight Status:")
display(df.sort_values(by=['Airline', 'Flight_Status']))
```

```python
print("Number of Flights by Airline:")
display(df.groupby('Airline').size())

print("Flights by Airline and Status:")
display(df.groupby(['Airline', 'Flight_Status']).size())

print("Flight Status Distribution by Airline:")
display(df.groupby('Airline')['Flight_Status'].value_counts())

print("Number of Flights by Status:")
display(df.groupby('Flight_Status').size())
```

```python
status_mapping = {
    'On Time': 1,
    'Delayed': 2,
    'Cancelled': 3
}

df['Status_Code'] = df['Flight_Status'].map(status_mapping)

df[['Flight_Status', 'Status_Code']].head()
```

```python
df['Airline_Status'] = df['Airline'] + ' - ' + df['Flight_Status']

df[['Airline', 'Flight_Status', 'Airline_Status']].head()
```

```python
df_transformed = df.copy()

df_transformed['Status_Code'] = df_transformed['Flight_Status'].map(status_mapping)

df_transformed.head()
```

```python
airline_counts = df['Airline'].value_counts()

print("Flights by Airline:")
display(airline_counts)

print("\nAirline with the Most Flights:")
print(airline_counts.idxmax())
```

```python
status_counts = df['Flight_Status'].value_counts()

print("Flight Status Distribution:")
display(status_counts)

print("\nMost Common Flight Status:")
print(status_counts.idxmax())
```

```python
delayed_flights = df[df['Flight_Status'] == 'Delayed']

print("Delayed Flights by Airline:")
display(delayed_flights['Airline'].value_counts())
```

```python
cancelled_flights = df[df['Flight_Status'] == 'Cancelled']

print("Cancelled Flights by Airline:")
display(cancelled_flights['Airline'].value_counts())
```

```python
delayed_counts = df[df['Flight_Status'] == 'Delayed']['Airline'].value_counts()

print("Airline with the Most Delayed Flights:")
print(delayed_counts.idxmax())
print("Number of Delayed Flights:", delayed_counts.max())
```

```python
cancelled_counts = df[df['Flight_Status'] == 'Cancelled']['Airline'].value_counts()

print("Airline with the Most Cancelled Flights:")
print(cancelled_counts.idxmax())
print("Number of Cancelled Flights:", cancelled_counts.max())
```

```python
print("Total Number of Flights:", len(df))

print("\nFlights by Airline:")
print(df['Airline'].value_counts())

print("\nFlight Status Distribution:")
print(df['Flight_Status'].value_counts())

print("\nDelayed Flights by Airline:")
print(df[df['Flight_Status'] == 'Delayed']['Airline'].value_counts())

print("\nCancelled Flights by Airline:")
print(df[df['Flight_Status'] == 'Cancelled']['Airline'].value_counts())
```

## Key Findings and Observations

1. The dataset contains a total of **180 flight records**.

2. **Vistara** is the airline with the highest number of flights, with **37 flights**.

3. **On Time** is the most common flight status, with **114 flights**.

4. **Vistara** has the highest number of delayed flights, with **13 delayed flights**.

5. **SpiceJet** has the highest number of cancelled flights, with **3 cancelled flights**.

6. Out of 180 flights, **59 flights were delayed** and **7 flights were cancelled**, while **114 flights operated on time**.

7. The analysis shows that **Vistara, SpiceJet, and IndiGo** have relatively higher numbers of flight records compared with the other airlines in the dataset.

8. Grouping and filtering the data using Pandas helped identify patterns in **flight operations, delays, and cancellations across different airlines**.
