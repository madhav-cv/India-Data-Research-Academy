```python
#Assignment: Simple Calculator (Day 1)
#Create a Python program that accepts two numbers from the user and performs all the basic arithmetic operations on them.
#Your program should display the results of addition (+), subtraction (-), multiplication (*), division (/), floor division (//), modulus (%), and exponentiation (**).
#Use appropriate variable names, convert user input to numeric data types, and display the output clearly using formatted print statements.


firstnum = float(input("Enter the first number: "))
secondnum = float(input("Enter the second number: "))


addition = firstnum + secondnum
subtraction = firstnum - secondnum
multiplication = firstnum * secondnum


print("\n Simple Calculator Results ")
print("Addition (+):", addition)
print("Subtraction (-):", subtraction)
print("Multiplication (*):", multiplication)


if secondnum != 0:
    division = firstnum / secondnum
    floor_division = firstnum // secondnum
    modulus = firstnum % secondnum

    print("Division (/):", division)
    print("Floor Division (//):", floor_division)
    print("Modulus (%):", modulus)
else:
    print("Division (/): Cannot divide by zero")
    print("Floor Division (//): Cannot divide by zero")
    print("Modulus (%): Cannot divide by zero")


exponentiation = firstnum ** secondnum
print("Exponentiation (**):", exponentiation)
```
